#### Create Game board

```python
def create_board():
    return [" " for _ in range(9)]
```

#### Display Board

```python
def display_board(board):
    print("\n")
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print("\n")
```

#### Player Move Function

```python
def player_move(board, player):
    while True:
        try:
            position = int(input(f"player {player}, choose position (1-9): ")) - 1
            if board[position] == " ":
                board[position] = player
                break
            else:
                print("Position already taken!")
        except:
            print("invalid input! Try again.")
```

#### Check Winner

```python
def check_winner(board, player):
    win_conditions= [
        [0,1,2], [3,4,5], [6,7,8],  #rows
        [0,3,6], [1,4,7], [2,5,8],  #columns
        [0,4,8], [2,4,6]            # diagonals
    ]

    for condition in win_conditions:
        if all(board[i] == player for i in condition):
            return True
    return False
```

#### Check Draw

```python
def check_draw(board):
    return " " not in board
```

#### Main game loop

```python
def play_game():
    board = create_board()
    current_player = "X"

    while True:
        display_board(board)
        
        move_status = player_move(board, current_player)

        if move_status == "exit":
            print("👋 Game exited!")
            return "exit"

        if check_winner(board, current_player):
            display_board(board)
            print(f" 🎉 player {current_player} wins!")
            break

        if check_draw(board):
            display_board(board)
            print(" 😮 It's a draw!")
            break

        current_player = "O" if current_player == "X" else "X"
```

#### Exit function

```python
def player_move(board, player):
    while True:
        position = input(f"player {player}, choose position (1-9) or type 'exit': ")

        if position.lower() == "exit":
            return "exit"

        try:
            position = int(position) - 1
            if board[position] == " ":
                board[position] = player
                return "continue"
            else:
                print(" ⚠ position already taken!")
        except:
            print(" ❌ invalid input! Try again.")
```

#### Play Game

```python
play_game()
```

#### Restart Game

```python
while True:
    result = play_game()

    if result == "exit":
        break

    again = input("Do you want to play again? (y/n): ")

    if again.lower() != 'y':
        print(" Thanks for Playing!")
        break
```
